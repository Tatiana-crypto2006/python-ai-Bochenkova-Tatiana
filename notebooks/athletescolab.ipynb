```python
# 🐱 Шаг 2A. Чтение CSV-файлов в pandas

import pandas as pd

df_athletes = pd.read_csv("data/athletes.csv")
df_awards = pd.read_csv("data/athletes_awards.csv")
df_bio = pd.read_csv("data/athletes_bio.csv")

print("✅ Загружено строк в df_athletes:", len(df_athletes))
print("✅ Загружено строк в df_awards:", len(df_awards))
print("✅ Загружено строк в df_bio:", len(df_bio))
```

```python
# Переименование столбцов для df_athletes
df_athletes = df_athletes.rename(columns={
    'athlete': 'athlete_uri',
    'athleteLabel': 'athlete_name',
    'sportLabel': 'sport',
    'genderLabel': 'gender',
    'countryLabel': 'country',
    'birthDate': 'birth_date',
    'deathDate': 'death_date',
    'height': 'height_cm',
    'weight': 'weight_kg'
})

# Проверяем новые названия столбцов
print("Столбцы в df_athletes:", df_athletes.columns.tolist())
```

```python
# Проверяем пропуски в df_athletes
print("Пропуски в df_athletes:")
print(df_athletes.isnull().sum())

# Удаляем строки с пропусками в критических столбцах
df_athletes = df_athletes.dropna(subset=['athlete_name', 'sport'])

# Заполняем пропуски в числовых столбцах средними значениями
df_athletes['height_cm'] = df_athletes['height_cm'].fillna(df_athletes['height_cm'].mean())
df_athletes['weight_kg'] = df_athletes['weight_kg'].fillna(df_athletes['weight_kg'].mean())

# Проверяем пропуски после обработки
print("\nПропуски после обработки:")
print(df_athletes.isnull().sum())
```

```python
# Приведение даты рождения к типу datetime
df_athletes['birth_date'] = pd.to_datetime(df_athletes['birth_date'], errors='coerce')

# Приведение роста и веса к числовому типу
df_athletes['height_cm'] = pd.to_numeric(df_athletes['height_cm'], errors='coerce')
df_athletes['weight_kg'] = pd.to_numeric(df_athletes['weight_kg'], errors='coerce')

# Проверяем типы данных
print("\nТипы данных в df_athletes:")
print(df_athletes.dtypes)
```

```python
# Сортировка по росту (по убыванию)
sorted_by_height = df_athletes.sort_values(by='height_cm', ascending=False)
print("Топ-5 самых высоких спортсменов:")
display(sorted_by_height.head(5))
```

```python
# Средний рост и вес по видам спорта
grouped_by_sport = df_athletes.groupby('sport').agg({
    'height_cm': 'mean',
    'weight_kg': 'mean'
}).reset_index()

print("\nСредний рост и вес по видам спорта:")
display(grouped_by_sport)
```

```python
# Количество спортсменов по полу и виду спорта
gender_sport_count = df_athletes.groupby(['sport', 'gender']).size().reset_index(name='count')
print("\nКоличество спортсменов по полу и виду спорта:")
display(gender_sport_count)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from matplotlib.lines import Line2D

# 1. Проверяем названия столбцов
print("Столбцы в DataFrame:", df_athletes.columns.tolist())

# 2. Определяем правильные имена столбцов
sport_col = 'sportLabel' if 'sportLabel' in df_athletes.columns else ('sport' if 'sport' in df_athletes.columns else 'sport_name')
gender_col = 'genderLabel' if 'genderLabel' in df_athletes.columns else ('gender' if 'gender' in df_athletes.columns else 'gender_label')

print(f"\nИспользуемые столбцы:")
print(f"  - sport_col: {sport_col}")
print(f"  - gender_col: {gender_col}")

# 3. Проверяем уникальные значения в столбце с полом
print(f"\nУникальные значения в столбце {gender_col}:", df_athletes[gender_col].unique())

# 4. Фильтруем данные (используем правильные значения для пола)
# Автоматически определяем значения пола
unique_genders = df_athletes[gender_col].unique()
male_female_values = []

# Определяем, какие значения являются мужскими и женскими
for val in unique_genders:
    val_lower = str(val).lower()
    if val_lower in ['male', 'мужской пол', 'мужчина', 'мужской', 'м', 'men', 'man']:
        male_female_values.append(val)
    elif val_lower in ['female', 'женский пол', 'женщина', 'женский', 'ж', 'women', 'woman']:
        male_female_values.append(val)

print(f"Отфильтрованные значения пола: {male_female_values}")

# Применяем фильтрацию
athletes_filtered = df_athletes[df_athletes[gender_col].isin(male_female_values)].copy()

# Удаляем выбросы
athletes_filtered = athletes_filtered[
    (athletes_filtered['height_cm'] >= 140) &
    (athletes_filtered['weight_kg'] <= 120)
]

print(f"\nПосле фильтрации: {len(athletes_filtered)} спортсменов")

# 5. Добавляем случайное смещение для наглядности
np.random.seed(42)
athletes_jittered = athletes_filtered.copy()
athletes_jittered['height_cm'] = athletes_jittered['height_cm'] + np.random.normal(0, 0.3, size=len(athletes_jittered))
athletes_jittered['weight_kg'] = athletes_jittered['weight_kg'] + np.random.normal(0, 0.3, size=len(athletes_jittered))

# 6. Создаём график
plt.figure(figsize=(14, 8))

# Получаем топ-10 видов спорта для читаемости
top_sports = athletes_jittered[sport_col].value_counts().head(10).index
athletes_top = athletes_jittered[athletes_jittered[sport_col].isin(top_sports)]

scatter = sns.scatterplot(
    data=athletes_top,
    x='height_cm',
    y='weight_kg',
    hue=sport_col,
    style=gender_col,
    s=50,
    alpha=0.6,
    edgecolor='black',
    linewidth=0.5
)

# 7. Настраиваем оформление
plt.title('Взаимосвязь роста и веса спортсменов по видам спорта и полу',
          fontsize=16, pad=20, fontweight='bold')
plt.xlabel('Рост (см)', fontsize=14, fontweight='bold')
plt.ylabel('Вес (кг)', fontsize=14, fontweight='bold')
plt.grid(True, linestyle='--', alpha=0.2)

# 8. Настраиваем легенду
plt.legend(
    title=f'{sport_col} / {gender_col}',
    bbox_to_anchor=(1.05, 1),
    loc='upper left',
    fontsize=10,
    frameon=True,
    fancybox=True,
    shadow=True
)

plt.tight_layout()
plt.show()
```

```python
from google.colab import files

# Загружаем все три файла
uploaded = files.upload()

# Проверяем, какие файлы были загружены
print("Загруженные файлы:", list(uploaded.keys()))
```

```python
import pandas as pd
import io

# Чтение файлов в DataFrame
athletes_df = pd.read_csv(io.BytesIO(uploaded['athletes.csv']))
awards_df = pd.read_csv(io.BytesIO(uploaded['athletes_awards.csv']))
bio_df = pd.read_csv(io.BytesIO(uploaded['athletes_bio.csv']))

# Проверяем первые строки каждого DataFrame
print("🔹 athletes.csv:")
display(athletes_df.head())

print("\n🔹 athletes_awards.csv:")
display(awards_df.head())

print("\n🔹 athletes_bio.csv:")
display(bio_df.head())
```

```python
# Пример переименования столбцов для athletes_df
athletes_df = athletes_df.rename(columns={
    'athlete': 'athlete_uri',
    'athleteLabel': 'athlete_name',
    'sportLabel': 'sport',
    'genderLabel': 'gender',
    'countryLabel': 'country',
    'birthDate': 'birth_date',
    'deathDate': 'death_date',
    'height': 'height_cm',
    'weight': 'weight_kg'
})

# Проверяем новые названия столбцов
print("Столбцы в athletes_df:", athletes_df.columns)
```

```python
# Проверяем пропуски в athletes_df
print("Пропуски в athletes_df:")
print(athletes_df.isnull().sum())

# Удаляем строки с пропусками в критических столбцах
athletes_df = athletes_df.dropna(subset=['athlete_name', 'sport'])

# Заполняем пропуски в числовых столбцах средними значениями
athletes_df['height_cm'] = athletes_df['height_cm'].fillna(athletes_df['height_cm'].mean())
athletes_df['weight_kg'] = athletes_df['weight_kg'].fillna(athletes_df['weight_kg'].mean())

# Проверяем пропуски после обработки
print("\nПропуски после обработки:")
print(athletes_df.isnull().sum())
```

```python
# Приведение даты рождения к типу datetime
athletes_df['birth_date'] = pd.to_datetime(athletes_df['birth_date'], errors='coerce')

# Приведение роста и веса к числовому типу
athletes_df['height_cm'] = pd.to_numeric(athletes_df['height_cm'], errors='coerce')
athletes_df['weight_kg'] = pd.to_numeric(athletes_df['weight_kg'], errors='coerce')

# Проверяем типы данных
print("\nТипы данных в athletes_df:")
print(athletes_df.dtypes)
```

```python
# Сортировка по росту (по убыванию)
sorted_by_height = athletes_df.sort_values(by='height_cm', ascending=False)
print("Топ-5 самых высоких спортсменов:")
display(sorted_by_height.head(5))
```

```python
# Средний рост и вес по видам спорта
grouped_by_sport = athletes_df.groupby('sport').agg({
    'height_cm': 'mean',
    'weight_kg': 'mean'
}).reset_index()

print("\nСредний рост и вес по видам спорта:")
display(grouped_by_sport)
```

```python
# Количество спортсменов по полу и виду спорта
gender_sport_count = athletes_df.groupby(['sport', 'gender']).size().reset_index(name='count')
print("\nКоличество спортсменов по полу и виду спорта:")
display(gender_sport_count)
```

```python
# Объединяем athletes_df и awards_df
merged_df = pd.merge(athletes_df, awards_df, on='athlete_uri', how='left')

# Объединяем результат с bio_df
merged_df = pd.merge(merged_df, bio_df, on='athlete_uri', how='left')

print("\nОбъединённая таблица (первые 5 строк):")
display(merged_df.head())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Фильтруем данные: оставляем только мужчин и женщин
athletes_filtered = athletes_df[athletes_df['gender'].isin(['мужской пол', 'женский пол'])]

# 2. Удаляем выбросы: рост < 140 см или вес > 120 кг
athletes_filtered = athletes_filtered[
    (athletes_filtered['height_cm'] >= 140) &
    (athletes_filtered['weight_kg'] <= 120)
]

# 3. Создаём график
plt.figure(figsize=(14, 8))

# 4. Рисуем Scatter Plot с цветом по виду спорта и стилем по полу
scatter = sns.scatterplot(
    data=athletes_filtered,
    x='height_cm',
    y='weight_kg',
    hue='sport',          # Цвет по виду спорта
    style='gender',      # Стиль по полу
    palette={'биатлон': 'red', 'плавание': 'blue', 'фигурное катание': 'green'},  # Явное задание цветов
    s=50,                # Уменьшаем размер точек
    alpha=0.5            # Увеличиваем прозрачность
)

# 5. Настраиваем оформление
plt.title('Взаимосвязь роста и веса спортсменов по видам спорта и полу', fontsize=16, pad=20)
plt.xlabel('Рост (см)', fontsize=14)
plt.ylabel('Вес (кг)', fontsize=14)
plt.grid(True, linestyle='--', alpha=0.3)

# 6. Настраиваем легенду
# Убираем автоматическую легенду
scatter.legend_.remove()

# Создаём кастомную легенду для вида спорта
from matplotlib.lines import Line2D
legend_elements_sport = [
    Line2D([0], [0], marker='o', color='w', label='Биатлон',
           markerfacecolor='red', markersize=10),
    Line2D([0], [0], marker='o', color='w', label='Плавание',
           markerfacecolor='blue', markersize=10),
    Line2D([0], [0], marker='o', color='w', label='Фигурное катание',
           markerfacecolor='green', markersize=10)
]

# Создаём кастомную легенду для пола
legend_elements_gender = [
    Line2D([0], [0], marker='o', color='w', label='Женский пол',
           markerfacecolor='none', markeredgecolor='black', markeredgewidth=1.5, markersize=10),
    Line2D([0], [0], marker='X', color='w', label='Мужской пол',
           markerfacecolor='none', markeredgecolor='black', markeredgewidth=1.5, markersize=10)
]

# Добавляем легенды на график
plt.legend(
    handles=legend_elements_sport + legend_elements_gender,
    title='Вид спорта / Пол',
    bbox_to_anchor=(1.05, 1),  # Перемещаем легенду вправо
    loc='upper left',
    fontsize=12
)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from matplotlib.lines import Line2D

# 1. Проверяем названия столбцов
print("Столбцы в DataFrame:", athletes_df.columns.tolist())

# 2. Определяем правильные имена столбцов
sport_col = 'sportLabel' if 'sportLabel' in athletes_df.columns else 'sport'
gender_col = 'genderLabel' if 'genderLabel' in athletes_df.columns else ('gender' if 'gender' in athletes_df.columns else 'gender_label')

# 3. Проверяем уникальные значения в столбце с полом
print("Уникальные значения в столбце с полом:", athletes_df[gender_col].unique())

# 4. Фильтруем данные (используем правильные значения для пола)
# Если значения на русском:
male_female_values = ['мужской пол', 'женский пол']
# Если значения на английском:
# male_female_values = ['male', 'female']

athletes_filtered = athletes_df[athletes_df[gender_col].isin(male_female_values)]
athletes_filtered = athletes_filtered[
    (athletes_filtered['height_cm'] >= 140) &
    (athletes_filtered['weight_kg'] <= 120)
]

# 5. Добавляем случайное смещение
np.random.seed(42)
athletes_jittered = athletes_filtered.copy()
athletes_jittered['height_cm'] = athletes_jittered['height_cm'] + np.random.normal(0, 0.3, size=len(athletes_jittered))
athletes_jittered['weight_kg'] = athletes_jittered['weight_kg'] + np.random.normal(0, 0.3, size=len(athletes_jittered))

# 6. Создаём график
plt.figure(figsize=(14, 8))
scatter = sns.scatterplot(
    data=athletes_jittered,
    x='height_cm',
    y='weight_kg',
    hue=sport_col,
    style=gender_col,
    palette={'биатлон': '#FF5733', 'плавание': '#33A1FF', 'фигурное катание': '#4CAF50'},
    s=30,
    alpha=0.6
)

# 7. Настраиваем оформление
plt.title('Взаимосвязь роста и веса спортсменов по видам спорта и полу', fontsize=16, pad=20, fontweight='bold')
plt.xlabel('Рост (см)', fontsize=14, fontweight='bold')
plt.ylabel('Вес (кг)', fontsize=14, fontweight='bold')
plt.grid(True, linestyle='--', alpha=0.2)

# 8. Настраиваем легенду
scatter.legend_.remove()
legend_elements_sport = [
    Line2D([0], [0], marker='o', color='w', label='Биатлон', markerfacecolor='#FF5733', markersize=10),
    Line2D([0], [0], marker='o', color='w', label='Плавание', markerfacecolor='#33A1FF', markersize=10),
    Line2D([0], [0], marker='o', color='w', label='Фигурное катание', markerfacecolor='#4CAF50', markersize=10)
]
legend_elements_gender = [
    Line2D([0], [0], marker='o', color='w', label='Женщины',
           markerfacecolor='none', markeredgecolor='black', markeredgewidth=1.5, markersize=10),
    Line2D([0], [0], marker='X', color='w', label='Мужчины',
           markerfacecolor='none', markeredgecolor='black', markeredgewidth=1.5, markersize=10)
]
plt.legend(
    handles=legend_elements_sport + legend_elements_gender,
    title='Вид спорта / Пол',
    bbox_to_anchor=(1.05, 1),
    loc='upper left',
    fontsize=12
)

plt.tight_layout()
plt.show()
```
